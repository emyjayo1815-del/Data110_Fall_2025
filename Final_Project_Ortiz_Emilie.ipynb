**Creating Map with Latitude and Longitude Fields Split by Gender**

```python
import pandas as pd
import folium
from folium.plugins import HeatMap

# Clean data
cleaned_df = df.dropna(subset=['Latitude', 'Longitude'])

# Filter MD only
md_df = cleaned_df[cleaned_df['State'] == 'MD'].copy()

# Convert Date column to datetime
md_df['Date Of Stop'] = pd.to_datetime(md_df['Date Of Stop'])

# Filter November 2024 + Warning (ALL)
base_filter = (
    (md_df['Date Of Stop'].dt.year == 2024) &
    (md_df['Date Of Stop'].dt.month == 11) &
    (md_df['Violation Type'].str.strip().str.lower() == 'warning')
)

all_warn = md_df[base_filter].copy()
male_warn = md_df[base_filter & (md_df['Gender'].str.strip().str.lower() == 'm')].copy()
female_warn = md_df[base_filter & (md_df['Gender'].str.strip().str.lower() == 'f')].copy()


# Center map using ALL warnings (so one consistent center)
center_lat = all_warn['Latitude'].median()
center_lon = all_warn['Longitude'].median()

m = folium.Map(location=[center_lat, center_lon], zoom_start=12)

# --- Layer 1: All warnings ---
fg_all = folium.FeatureGroup(name="All Warnings (Nov 2024)", show=True)
HeatMap(all_warn[['Latitude', 'Longitude']].values.tolist()).add_to(fg_all)
fg_all.add_to(m)

# --- Layer 2: Male warnings ---
fg_male = folium.FeatureGroup(name="Male Warnings (Nov 2024)", show=False)
HeatMap(male_warn[['Latitude', 'Longitude']].values.tolist()).add_to(fg_male)
fg_male.add_to(m)

# --- Layer 3: Female warnings ---
fg_female = folium.FeatureGroup(name="Female Warnings (Nov 2024)", show=False)
HeatMap(female_warn[['Latitude', 'Longitude']].values.tolist()).add_to(fg_female)
fg_female.add_to(m)

# Add layer toggle control
folium.LayerControl(collapsed=False).add_to(m)

# Save single HTML with all 3 layers
m.save("Warnings_MD_Nov2024_Heatmaps.html")
```

Creating Map with Latitude and Longitude Fields Split White Race and Gender

```python
import pandas as pd
import folium
from folium.plugins import HeatMap

# Clean data
cleaned_df = df.dropna(subset=['Latitude', 'Longitude'])

# Filter MD only
md_df = cleaned_df[cleaned_df['State'] == 'MD'].copy()

# Convert Date column to datetime
md_df['Date Of Stop'] = pd.to_datetime(md_df['Date Of Stop'])

# Base filter: Nov 2024 + Warning
base_filter = (
    (md_df['Date Of Stop'].dt.year == 2024) &
    (md_df['Date Of Stop'].dt.month == 11) &
    (md_df['Violation Type'].str.strip().str.lower() == 'warning')
)

# Apply base filter once
warn_df = md_df[base_filter].copy()

# Clean Race/Gender fields for consistent matching
warn_df['Race_clean'] = warn_df['Race'].astype(str).str.strip().str.lower()
warn_df['Gender_clean'] = warn_df['Gender'].astype(str).str.strip().str.lower()

# Get all race categories dynamically (your 5 races will appear here)
races = sorted(warn_df['Race_clean'].dropna().unique())

# Center map using ALL warnings (so one consistent center)
center_lat = warn_df['Latitude'].median()
center_lon = warn_df['Longitude'].median()

m = folium.Map(location=[center_lat, center_lon], zoom_start=12)

# Optional: layer for ALL warnings (all races, all genders)
fg_all = folium.FeatureGroup(name="All Warnings (All races) - Nov 2024", show=True)
HeatMap(warn_df[['Latitude', 'Longitude']].values.tolist()).add_to(fg_all)
fg_all.add_to(m)

# Create layers for each race and gender
for race in races:
    race_title = race.title()

    # Race: all genders
    df_race_all = warn_df[warn_df['Race_clean'] == race]
    if not df_race_all.empty:
        fg = folium.FeatureGroup(name=f"{race_title} - All (Nov 2024)", show=False)
        HeatMap(df_race_all[['Latitude', 'Longitude']].values.tolist()).add_to(fg)
        fg.add_to(m)

    # Race: male
    df_race_m = warn_df[(warn_df['Race_clean'] == race) & (warn_df['Gender_clean'] == 'm')]
    if not df_race_m.empty:
        fg = folium.FeatureGroup(name=f"{race_title} - Male (Nov 2024)", show=False)
        HeatMap(df_race_m[['Latitude', 'Longitude']].values.tolist()).add_to(fg)
        fg.add_to(m)

    # Race: female
    df_race_f = warn_df[(warn_df['Race_clean'] == race) & (warn_df['Gender_clean'] == 'f')]
    if not df_race_f.empty:
        fg = folium.FeatureGroup(name=f"{race_title} - Female (Nov 2024)", show=False)
        HeatMap(df_race_f[['Latitude', 'Longitude']].values.tolist()).add_to(fg)
        fg.add_to(m)

# Add layer toggle control
folium.LayerControl(collapsed=False).add_to(m)

# Save single HTML with all layers
m.save("Warnings_MD_Nov2024_AllRaces_GenderLayers.html")
```
