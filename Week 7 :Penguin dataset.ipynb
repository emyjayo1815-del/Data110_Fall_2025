**# Task 3: Make 3 subplots — scatter + regression — for each species separately**

```python
import matplotlib.pyplot as plt

# Filter by species
penguins_Adelie = penguins[penguins['species'] == 'Adelie']
penguins_Chinstrap = penguins[penguins['species'] == 'Chinstrap']
penguins_Gentoo = penguins[penguins['species'] == 'Gentoo']

# Create subplots (1 row, 3 columns)
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

# Adelie
axes[0].scatter(
    penguins_Adelie['bill_length_mm'],
    penguins_Adelie['bill_depth_mm']
)
axes[0].set_title('Adelie')
axes[0].set_xlabel('Bill Length (mm)')
axes[0].set_ylabel('Bill Depth (mm)')

# Regression line (no scatter points duplicated)
sns.regplot(
    data=penguins_Adelie,
    x="bill_length_mm",
    y="bill_depth_mm",
    scatter=False,
    ax=axes[0],
    color="red"
)


# Chinstrap
axes[1].scatter(
    penguins_Chinstrap['bill_length_mm'],
    penguins_Chinstrap['bill_depth_mm']
)
axes[1].set_title('Chinstrap')
axes[1].set_xlabel('Bill Length (mm)')

# Regression line (no scatter points duplicated)
sns.regplot(
    data=penguins_Chinstrap,
    x="bill_length_mm",
    y="bill_depth_mm",
    scatter=False,
    ax=axes[1],
    color="red"
)


# Gentoo
axes[2].scatter(
    penguins_Gentoo['bill_length_mm'],
    penguins_Gentoo['bill_depth_mm']
)
axes[2].set_title('Gentoo')
axes[2].set_xlabel('Bill Length (mm)')

# Regression line (no scatter points duplicated)
sns.regplot(
    data=penguins_Gentoo,
    x="bill_length_mm",
    y="bill_depth_mm",
    scatter=False,
    ax=axes[2],
    color="red"
)



plt.suptitle('Bill Length vs Bill Depth by Species')
plt.tight_layout()
plt.show()
```
