**Part 1: Understanding the Dataset**

```python
import seaborn as sns
import pandas as pd

# Load Titanic dataset
titanic = sns.load_dataset('titanic')

# View the first few rows of the dataset
titanic.head()
```

The Dataset is showing a detail list and a summary of the people that ride the Titanic and also those that survived or passed away after the crash. It also indicate the type of gender what level of class they had and the payment of each fare.

**Part 2: Fare and Survival Analysis**

Task 2: Fare Distribution

```python
import matplotlib.pyplot as plt
plt.hist (titanic['fare'], bins=10, color='skyblue', edgecolor='black')
plt.title('Fare Distribution of Titanic Passengers')
plt.xlabel('Fare ($)')
plt.show ()
```

Task 3: Fare and Survival Rates

```python
sns.kdeplot (titanic[titanic['survived'] == 1]['fare'], label='Survived', shade=True)
sns.kdeplot (titanic[titanic['survived'] == 0]['fare'], label='Not Survived', shade=True)
plt.show()
```

**Part 3: Class, Fare, and Survival Insights**

Task 4: Passenger Class, Fare, and Survival

```python
sns.kdeplot (titanic[titanic['survived'] == 1]['pclass'], label='Survived', shade=True)
sns.kdeplot (titanic[titanic['survived'] == 0]['pclass'], label='Not Survived', shade=True)
plt.show()
```

```python
plt.hist (titanic['pclass'], bins=10, color='skyblue', edgecolor='black')
plt.title('PClass Distribution of Titanic Passengers')
plt.xlabel('Fare ($)')
plt.show ()
```

Part 4: Age and Survival

**Task 5: Age of Survivors and Non-Survivors**

```python
sns.kdeplot (titanic[titanic['survived'] == 1]['age'], label='Survived', shade=True)
sns.kdeplot (titanic[titanic['survived'] == 0]['age'], label='Not Survived', shade=True)
plt.show()
```

Task 6: Pyramid Histogram for Age Survival by Sex

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create age bins
bins = np.arange(0, 80, 5)

# Filter data for male and female passengers
male_ages = titanic[titanic['sex'] == 'male']['age'].dropna()
female_ages = titanic[titanic['sex'] == 'female']['age'].dropna()

# Calculate histograms for male and female ages
male_hist, _ = np.histogram(male_ages, bins=bins)
female_hist, _ = np.histogram(female_ages, bins=bins)

# Create the pyramid histogram
plt.figure(figsize=(10, 7))
plt.barh(bins[:-1], male_hist, height=5, label='Male', color='skyblue')
plt.barh(bins[:-1], -female_hist, height=5, label='Female', color='salmon') # Use negative values for the female side

plt.xlabel('Number of Passengers')
plt.ylabel('Age')
plt.title('Age Distribution by Sex (Pyramid Histogram)')
plt.legend()
plt.grid(axis='x')
plt.show()
```
